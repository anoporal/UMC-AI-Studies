Primeiro, baixando o dataset do kaggle.

```python
# Importando o kagglehub para ter acesso direto ao dataset
import kagglehub

# Determinando o path do download do dataset
path = kagglehub.dataset_download("uciml/breast-cancer-wisconsin-data")

# Retornando ao usuário a confirmação do acesso do dataset via path definido
print("Path to dataset files:", path)
```

Agora, importando as bibliotecas.

```python
import pandas as pd # para criação e manipulação de dataframes
import matplotlib.pyplot as plt # para plotagem de gráficos
from sklearn.model_selection import train_test_split, cross_val_score # train_test_split: para separação da proporção de treino e teste | cross_val_score: para validação cruzada
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, ConfusionMatrixDisplay # accuracy_score: | classification_report: | confusion_matrix: | confusion_matrix_display:
from sklearn.preprocessing import StandardScaler # para equilíbrio da avaliação dos dados pelo algoritmo
from sklearn.pipeline import Pipeline # para recursos de Pipeline no desenvolvimento do SVM
from sklearn.svm import SVC # para recursos de SVM do tipo de classificação (SVC)
from sklearn.tree import DecisionTreeClassifier # para recursos da árvore de decisão (também do tipo de classificação)
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier # para recursos da floresta aleatória e XGBoost (também do tipo de classificação)
import os # para manipulação do path inserido
```

```python
# Definição dos dados do dataset em um dataframe
df_cancer = pd.read_csv(os.path.join(path, "data.csv"))

# Retirando a coluna 'id' (pois é desnecessária)
df = df_cancer.drop(columns=['id'])

# Retirando todas as colunas vazias
df = df.dropna(how='all', axis='columns')

# Retornando as 5 primeiras linhas do dataset
df.head()
```

Então, amostro os dados coletados com seus tipos, contagens e estatísticas, assim como a quantidade de dados nulos e duplicatas.

```python
# Retornando o formato do dataset (linhas x colunas)
print("Formato do Dataset:", df.shape)

# Retorno a quantidade de duplicatas, por coluna
print("\nQuantidade de duplicatas, por coluna:", df.duplicated().sum())

# Retorno a quantidade de valores faltantes, por coluna
print("\n-=-=-=-=-=-=-=-=QUANTIDADE DE DADOS FALTANTES-=-=-=-=-=-=-=-=\n", df.isnull().sum())

# Retornando as informações das colunas (nome da coluna, contagem de valores não-nulos e tipo da variável)
print("\n-=-=-=-=-=-=-=-=INFORMAÇÕES DAS COLUNAS-=-=-=-=-=-=-=-=")
df.info()

# Retornando a contagem dos valores por diagnóstico (quantidade de tumores benignos x malignos)
print("\n-=-=-=-=-=-=-=-=CONTAGEM POR DIAGNÓSTICO-=-=-=-=-=-=-=-=")
print(df['diagnosis'].value_counts())

# Retornando a descrição geral do dataset: contagem dos valores, média, desvio padrão, valor mínimo/máximo, Q1, Q2 e Q3
print("\n-=-=-=-=-=-=-=-=DESCRIÇÃO DO DATASET-=-=-=-=-=-=-=-=")
display(df.describe())
```

```python
# Retirando a coluna de diagnóstico (nossa variável-alvo) e retornando histogramas que mostram a distribuição das variáveis
df.drop(columns=['diagnosis']).hist(figsize=(10, 8))

plt.suptitle("Distribuição de variáveis")
plt.tight_layout()
plt.show()
```

```python
# Criando um dict com todos os 7 modelos que serão aplicados
# (todos os SVM + Árvore de Decisão + Floresta Aleatória + XGBoost)
models = {
    "SVM (Linear)": Pipeline([('scaler', StandardScaler()), ('svc', SVC(kernel='linear', random_state=42))]),
    "SVM (RBF)": Pipeline([('scaler', StandardScaler()), ('svc', SVC(kernel='rbf', random_state=42))]),
    "SVM (Polynomial)": Pipeline ([('scaler', StandardScaler()), ('svc', SVC(kernel='poly', random_state=42))]),
    "SVM (Sigmoid)": Pipeline([('scaler', StandardScaler()), ('svc', SVC(kernel='sigmoid', random_state=42))]),
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "Random Forest": RandomForestClassifier(n_estimators=100, random_state=42),
    "Gradient Boosting": GradientBoostingClassifier(n_estimators=100, random_state=42)
}
```

Agora então, separando as variáveis de teste e as de treino

```python
# Definindo as variáveis de X e y
X = df.drop(columns='diagnosis') # X = características
y = df['diagnosis'] # y = rótulo (variável-alvo)

# Dividindo os dados de treino e teste em 80% e 20%, respectivamente
X_train, X_test, y_train, y_test = train_test_split(
    X, y, train_size=0.8, test_size=0.2, random_state=42, stratify=y
)
```

```python
# Criando um dict para armazenar os resultados
results = {}

for name, model in models.items():
  # Validação cruzada com 5 folds
  cv_scores = cross_val_score(model, X_train, y_train, cv=5, scoring='accuracy')

  # Treinando o modelo
  model.fit(X_train, y_train)

  # Predição no conjunto de teste
  y_pred = model.predict(X_test)
  test_acc = accuracy_score(y_test, y_pred)

  results[name] = {
      'CV Mean Accuracy': cv_scores.mean(),
      'Test Accuracy': test_acc
  }

  print(f"=-=-=-=-=-{name}-=-=-=-=-=")
  print(f"Acurácia média: {cv_scores.mean():.4f}")
  print(f"Acurácia do teste: {test_acc:.4f}")
  print(classification_report(y_test, y_pred))

  # Retornando a matriz de confusão
  conf_matrix = confusion_matrix(y_test, y_pred)
  disp = ConfusionMatrixDisplay(confusion_matrix=conf_matrix, display_labels=model.classes_)
  disp.plot(cmap=plt.cm.BuPu)
  plt.title(f"Matriz de Confusão - {name}")
  plt.show()
  print("-"*50)
```

Então, comparando todos os modelos:

```python
# Tabela de resumo de comparação
df_results = pd.DataFrame(results).T.sort_values(by='Test Accuracy', ascending=False)
print("Resumo Comparativo de Desempenho:")
print(df_results)
```

Sendo assim, o modelo de SVM com kernel RBF obteve a maior acurácia, com 97.3% de acurácia nos testes e de acurácia média.

Separando o X (características) e o y (rótulos)

```python
# Separando os dados numéricoos em uma variável
col_num = ["price", "vintage_quality", "winemaking_score", "region_prestige",
           "region_terroir", "scarcity", "oak_months", "residual_sugar",
           "alcohol", "true_quality", "blind_rating_expert",
           "blind_rating_novice", "blind_rating", "sighted_rating"]

# Separando os dados categóricos em uma variável
col_cat = ["region", "grape"]

# Definindo X como todas as colunas (características)
X = df[col_num + col_cat]
# Definindo a variável-alvo (rótulo), que é a variável 'overpriced'
y = df["overpriced"]

# Retornando as proporções das variáveis
print("Tamanho de X:", X.shape)
print("Tamanho de y:", y.shape)
print("\nProporção das classes:")
print(y.value_counts(normalize=True)*100)
```

Então, seguindo para o treinamento e determinação dos modelos

```python
X_encoded = pd.get_dummies(X, columns=col_cat, drop_first=True)
X_encoded = X_encoded.fillna(X_encoded.median())

X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y, test_size=0.2, random_state=42, stratify=y
)

models = {
    "SVM (Linear)": Pipeline([('scaler', StandardScaler()), ('svc', SVC(kernel='linear', random_state=42))]),
    "SVM (RBF)": Pipeline([('scaler', StandardScaler()), ('svc', SVC(kernel='rbf', random_state=42))]),
    "SVM (Polynomial)": Pipeline([('scaler', StandardScaler()), ('svc', SVC(kernel='poly', random_state=42))]),
    "SVM (Sigmoid)": Pipeline ([('scaler', StandardScaler()), ('svc', SVC(kernel='sigmoid', random_state=42))]),
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "Random Forest": RandomForestClassifier(n_estimators=100, random_state=42),
    "Gradient Boosting": GradientBoostingClassifier(n_estimators=100, random_state=42)
}
```

Agora, determinando as matrizes de confusão dos modelos

```python
results = {}

for name, model in models.items():
  # Validação cruzada (5 folds)
  cv_scores = cross_val_score(model, X_train, y_train, cv=5, scoring='accuracy')

  model.fit(X_train, y_train)
  y_pred = model.predict(X_test)
  test_acc = accuracy_score(y_test, y_pred)

  results[name] = {
      'Acurácia média (CV)': cv_scores.mean(),
      'Acurácia Teste': test_acc
  }

  print(f"{name}")
  print(f"Acurácia média (CV): {cv_scores.mean():.4f}")
  print(f"Acurácia nos testes: {test_acc:.4f}\n")
  print(classification_report(y_test, y_pred))

  cm = confusion_matrix(y_test, y_pred)
  disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=model.classes_)
  disp.plot(cmap=plt.cm.BuPu)
  plt.title(f"Matriz de Confusão - {name}")
  plt.show()
  print("-"*50)
```

```python
df_results = pd.DataFrame(results).T.sort_values(by='Acurácia Teste', ascending=False)
print("Resumo Comparativo de Desempenho:")
print(df_results)
```

```python
# Versão 1: Usando NDArray da biblioteca numpy.typing (recomendado e mais limpo)
def create_columns_index(first: int, size: int) -> NDArray[uint8]:
    # Gera a sequência contínua de índices numéricos de colunas em baixo nível
    return np.array(range(size), dtype=uint8) + first

# Fatia a aba do Excel em sub-tabelas utilizando o método nativo em C 'take()' do Pandas
def tables_to_df(sheet: DataFrame, indexes_sizes: list[tuple[int, int]]) -> list[DataFrame]:
    return [
        # Aplica a extração de alta performance para cada tupla (índice, tamanho) da aba
        sheet.take(indices=create_columns_index(index, size), axis='columns')
        for index, size in indexes_sizes
    ]
```

```python
class Votes(ABC):
    __slots__ = ('_results', '_electoral_votes', '_total_per_state')

    _results: DataFrame
    _electoral_votes: NDArray[np.uint8]
    _total_per_state: NDArray[np.uint32]

    def _clean_lines(self, df: DataFrame) -> DataFrame:
        # Método interno de higienização: Remove linhas completamente vazias
        return df.dropna(axis='index', how='all')

    def _series_to_nparray(self, df: DataFrame, dtype: DTypeLike = np.uint16) -> NDArray:
        df = self._clean_lines(df)
        return np.asarray(df[df.columns[0]].to_numpy(), dtype)

    def _get_totals(self, df: DataFrame) -> NDArray[np.uint32]:
        return df.values[-1]

    def get_electoral_votes(self) -> NDArray[np.uint8]:
        return self._electoral_votes[:-1]

    def get_total_per_state(self) -> NDArray[np.uint32]:
        return self._total_per_state[:-1]

    def get_results(self) -> DataFrame:
        return self._results[:-1]

    def get_total_results(self) -> NDArray[np.uint32]:
        return self._get_totals(self._results)

    def _set_header_from(self, df: DataFrame, line_index: int = 0) -> DataFrame:
        df.columns = df.values[line_index]
        return df.drop(index=line_index)

    def _discard_lines(self, source: DataFrame, indexes: list[int]) -> DataFrame:
        return source.drop(index=[source.index[index] for index in indexes])

    def _prepare_dataframe(self, df: DataFrame, header_index: int = -1, discard_indexes: list[int] = [], clean_lines: bool = True) -> DataFrame:
        if clean_lines:
            df = self._clean_lines(df)
        if header_index != -1:
            df = self._set_header_from(df, header_index)
        if not len(discard_indexes) == 0:
            df = self._discard_lines(df, discard_indexes)
        return df

    def _prepare_nparray(self, df: DataFrame, header_index: int = -1, discard_indexes: list[int] = [], clean_lines: bool = True, dtype: DTypeLike = np.uint32) -> NDArray:
        return self._series_to_nparray(self._prepare_dataframe(df, header_index, discard_indexes, clean_lines), dtype)
```

```python
# Puxa o DataFrame com as contagens de votos previstos na pesquisa de 1924
cand_df_1924 = votes1924.get_candidates()

# Extrai os vetores de votos de Coolidge e Davis
# Coolidge (Republicano - Coluna 0)
# Davis (Democrata - Coluna 1)
coolidge = cand_df_1924[cand_df_1924.columns[0]]
davis = cand_df_1924[cand_df_1924.columns[1]]

# Proporção Prevista (Two-Party Share da Pesquisa)
# Calcula a porcentagem do candidato Republicano em relação ao total bipartidário (0% a 100%)
# Isola os dois partidos para o deslocamento eleitoral
# Usamos um candidato como referência, pois os dois representam 100%
expected_proportions_1924 = coolidge / (coolidge + davis) * 100
print("Proporção Prevista (Pesquisa):")
print(expected_proportions_1924)
```

```python
# Puxa o DataFrame oficial dos resultados eleitorais de 1924
results = votes1924.get_results()

# Extrai a contagem real de votos das urnas para Coolidge e Davis
coolidge_expected = results[results.columns[1]]
davis_expected = results[results.columns[4]]

# Proporção Esperada (Urnas Reais em %)
# Ao multiplicar por 100 estou passando para porcentagem
# Usa os dados reais da eleição
actual_proportions_1924 = (coolidge_expected / ( coolidge_expected + davis_expected)) * 100
print("\nProporção Esperada (Urnas):")
print(actual_proportions_1924)
```

```python
models = {
    "SVM (Linear)": Pipeline([('scaler', StandardScaler()), ('svc', SVC(kernel='linear', random_state=42))]),
    "SVM (RBF)": Pipeline([('scaler', StandardScaler()), ('svc', SVC(kernel="rbf", random_state=42))]),
    "SVM (Polynomial)": Pipeline([('scaler', StandardScaler()), ('svc', SVC(kernel='poly', random_state=42))]),
    "SVM (Sigmoid)": Pipeline([('scaler', StandardScaler()), ('svc', SVC(kernel='sigmoid', random_state=42))]),
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "Random Forest": RandomForestClassifier(n_estimators=100, random_state=42),
    "Gradient Boosting": GradientBoostingClassifier(n_estimators=100, random_state=42)
}

results = {}

for name, model in models.items():
  cv_scores = cross_val_score(model, X_train, y_train, cv=5, scoring='accuracy')
  model.fit(X_train, y_train)
  y_pred = model.predict(X_test)

  acc = accuracy_score(y_test, y_pred)

  results[name] = {
      'Acurácia Média (CV)': cv_scores.mean(),
      'Acurácia Teste': acc
  }

  print(f"=-=-=-=-=-=-=-=-=-={name}-=-=-=-=-=-=-=-=-=-")
  print(f"Acurácia Média (CV): {cv_scores.mean()*100:.3f}% | Acurácia Teste: {acc*100:.3f}%")

  cm = confusion_matrix (y_test, y_pred, labels=model.classes_)
  fig, ax = plt.subplots(figsize=(6, 5))
  disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=model.classes_)
  disp.plot(cmap=plt.cm.BuPu, ax=ax)
  plt.title(f"Matriz de Confusão - {name}")
  plt.show()
  print("-"*60+"\n")
```

Por fim, criando um resumo comparativo das métricas de acurácia

```python
df_resume = pd.DataFrame(results).T.sort_values(by='Acurácia Média (CV)', ascending=False)
print("Resumo comparativo das métricas de acurácia")
display(df_resume)
```
